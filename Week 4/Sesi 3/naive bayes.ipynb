```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris

# Load dataset Iris dari scikit-learn
iris = load_iris()
```

##### Penjelasan:
Kode ini digunakan untuk mengimpor library yang diperlukan. numpy digunakan untuk pengolahan array, pandas untuk mengolah data dalam bentuk DataFrame, sedangkan load_iris digunakan untuk mengambil dataset Iris dari Scikit-learn.

load dataset Iris dari scikit-learn: Kode ini mengambil dataset Iris yang sudah tersedia di Scikit-learn.

```python
# Konversi data menjadi DataFrame Pandas
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]

# Definisikan X (fitur) dan y (target/label)
X = df.drop(columns=["Species"])
y = df["Species"]

print(X)
```

##### Penjelasan:
Konversi data menjadi DataFrame Pandas: Data fitur Iris diubah menjadi DataFrame dan ditambahkan kolom Species sebagai target atau jenis bunga Iris.

X (fitur) dan y (target/label): X berisi empat fitur bunga, yaitu sepal length, sepal width, petal length, dan petal width.

```python
y
```

##### Penjelasan:
 y berisi jenis bunga iris

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

##### Penjelasan:
Label jenis bunga yang awalnya berupa teks diubah menjadi angka agar dapat digunakan oleh model. setosa, versicolor, dan virginica akan direpresentasikan sebagai label numerik.

```python
from sklearn.model_selection import train_test_split

X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

##### Penjelasan:
Dataset dibagi menjadi 80% data training dan 20% data testing. random_state=0 digunakan agar pembagian data tetap sama ketika kode dijalankan kembali.

```python
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

##### Penjelasan:
GaussianNB() membuat model Gaussian Naïve Bayes. Kemudian fit() digunakan untuk melatih model menggunakan data training.

```python
result = NB_model.predict(X_test)
result
```

##### Penjelasan:
Kode tersebut digunakan untuk memprediksi jenis bunga berdasarkan data testing.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

##### Penjelasan:
accuracy_score() digunakan untuk mengetahui tingkat ketepatan model dalam memprediksi data testing.

```python
#prediksi inputan data baru
#input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

##### Penjelasan:
Kode ini digunakan untuk memprediksi jenis bunga dari data baru menggunakan model Gaussian Naïve Bayes. Data [5.1, 3.5, 1.4, 0.2] dimasukkan ke model, kemudian hasil prediksi yang berupa angka diubah kembali menjadi nama species menggunakan inverse_transform().

```python
input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)

result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

##### Penjelasan:
Kode ini digunakan untuk memprediksi jenis bunga dari data baru dan menampilkan hasilnya dalam bentuk nama species

```python
input_data = pd.DataFrame(np.array([[7.3, 3.8, 4.6, 2.3]]), columns=iris.feature_names)

result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

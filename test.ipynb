```python
#operasi elementwise (mengoperasikan setiap elemen yang berkorespondensi dengan elemen yang sama)
a = np.array([1, 2, 3, 4, 5])
b = np.array([6, 7, 8, 9, 10])

hasil1 = a+b
hasil2 = a-b
hasil3 = a*b
hasil4 = a/b
hasil5 = a**2

print("hasil 1 =", hasil1)
print("hasil 2 =", hasil2)
print("hasil 3 =", hasil3)
print("hasil 4 =", hasil4)
print("hasil 5 =", hasil5)
```

```python
#invers matriks
a = np.array( [(1,-1), (1,1)] )
print(a)

a_inv = np.linalg.inv(a)
print(a_inv)

det_a = np.linalg.det(a)
print(det_a)
```

```python
#solusi persamaan linear
A = np.array( [(1,-6),(2,-5)] )
Y = np.array([-35, 21])
print(A)
print(Y)

A_inv = np.linalg.inv(A)

#menyelesaikan persamaan linear
X1 = np.dot(A_inv, Y)
print(X1)

#cara lain
X2 = np.linalg.solve(A,Y)
print(X2)
```

```python
def f(x,y):
    return np.cos(np.exp(-x**2-y**2))*np.sin(np.sqrt(x**2+y**2))

x = np.linspace(-np.pi, np.pi, 30)
y = np.linspace(-np.pi, np.pi, 30)

X,Y = np.meshgrid(x,y)
Z = f(X,Y)

fig = plt.figure()
ax = plt.axes(projection = '3d')
ax.contour3D(X, Y, Z, 50, cmap = 'binary')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z');
```

```python
#Tugas 02
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 2, 500)
y = np.linspace(-2, 2, 500)

X, Y = np.meshgrid(x, y)

Z = (X**6 
     + 3*X**4*Y**2 
     - 3*X**4 
     + 3*X**2*Y**4 
     - X**2*Y**3 
     - 6*X**2*Y**2 
     + 3*X**2 
     + Y**6 
     - 3*Y**4 
     + 3*Y**2)

plt.figure(figsize=(8, 8))
plt.contour(X, Y, Z, levels=[1])

plt.xlabel('x')
plt.ylabel('y')
plt.title('Grafik 2 Dimensi')
plt.grid()
plt.axis('equal')

plt.show()
```

```python
#Tugas 03
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)

X, Y = np.meshgrid(x, y)

Z = ((5 * Y / 4) - np.abs(X))**2 + X**2

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

ax.set_title('Grafik 3 Dimensi')

plt.show()
```
